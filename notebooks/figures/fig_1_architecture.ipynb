```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

mpl.rcParams.update({
    "font.family": "sans-serif",
    "font.sans-serif": ["Arial", "Helvetica", "DejaVu Sans"],
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
    "svg.fonttype": "none",
})


# ----------------------------
# Settings
# ----------------------------
np.random.seed(7)

CELL_COLOR = "#4C72B0"
GENE_COLOR = "#8A3A3A"
BG_COLOR = "#C9BDB1"

# ----------------------------
# Define 3 distinct cell-island centers
# ----------------------------
centers = np.array([
    [-1.589,  3.309],
    [-1.491, -2.130],
    [ 3.653, -0.362],
])

n_cells_per_cluster = [220, 220, 220]
cell_spreads = [0.42, 0.42, 0.42]

# ----------------------------
# Generate compact cell clusters
# ----------------------------
cell_points = []
for (cx, cy), n, s in zip(centers, n_cells_per_cluster, cell_spreads):
    pts = np.random.normal(loc=(cx, cy), scale=s, size=(n, 2))
    cell_points.append(pts)
cell_points = np.vstack(cell_points)

# ----------------------------
# Generate genes
# Mostly compact in the middle, with a few around clusters
# ----------------------------
gene_points = []

# Main central compact gene cloud
central_genes = np.random.normal(
    loc=(0.0, 0.1),
    scale=(0.65, 0.55),
    size=(500, 2)
)
gene_points.append(central_genes)

# Extra very compact core to make the middle denser
core_genes = np.random.normal(
    loc=(0.0, 0.1),
    scale=(0.28, 0.22),
    size=(45, 2)
)
gene_points.append(core_genes)

# A few genes near each cell island
n = 50
local_gene_counts = [n, n, n]
for (cx, cy), n in zip(centers, local_gene_counts):
    pts = np.random.normal(loc=(cx, cy), scale=0.65, size=(n, 2))
    gene_points.append(pts)

gene_points = np.vstack(gene_points)

# ----------------------------
# Plot
# ----------------------------
fig, ax = plt.subplots(figsize=(2.8, 2.1), facecolor="white")

# Genes first
ax.scatter(
    gene_points[:, 0], gene_points[:, 1],
    s=3, c=GENE_COLOR, alpha=1, linewidths=0.6, zorder=3
)

# Cells on top
ax.scatter(
    cell_points[:, 0], cell_points[:, 1],
    s=3, c=CELL_COLOR, alpha=1, linewidths=0.3, zorder=2
)

ax.set_aspect("equal")
ax.axis("off")
ax.margins(0.12)
plt.savefig(outdir / 'panel_c_embedding.svg', bbox_inches='tight', pad_inches=0.02)
plt.show()
plt.tight_layout()
plt.show()
```

## Formulas

```python
from pathlib import Path
import html
import re

OUTDIR = Path("output/fig_1")
OUTDIR.mkdir(parents=True, exist_ok=True)

FONT_SIZE = 5.5  # pt
SUB_SIZE = FONT_SIZE * 0.7
TEXT_COLOR = "#222222"

ARIAL = "Arial, Helvetica, sans-serif"
SYMBOL = "Symbol"

BASE_Y = 13.0
SUB_DY = 2.4
SUP_DY = -3.2


def esc(s):
    return html.escape(s, quote=False)


def estimate_width(text, size=FONT_SIZE):
    """Approximate SVG text advance in pt units."""
    width = 0
    for ch in text:
        if ch == " ":
            width += 0.30 * size
        elif ch in "ijl.,:;()[]{}":
            width += 0.33 * size
        elif ch in "+-=−~·/|":
            width += 0.50 * size
        elif re.match(r"[A-Z]", ch):
            width += 0.68 * size
        else:
            width += 0.56 * size
    return width


def text_svg(x, y, content, font=ARIAL, size=FONT_SIZE, weight="normal", label=None):
    label_attr = f' inkscape:label="{esc(label)}"' if label else ""
    return (
        f'  <text x="{x:.2f}" y="{y:.2f}"{label_attr} '
        f'xml:space="preserve" '
        f'style="font-family:{font};font-size:{size:.2f}pt;'
        f'font-weight:{weight};fill:{TEXT_COLOR};">'
        f'{esc(content)}</text>\n'
    )


def part(content, font=ARIAL, size=FONT_SIZE, weight="normal", kind="base"):
    return {
        "content": content,
        "font": font,
        "size": size,
        "weight": weight,
        "kind": kind,
    }


def latin(s):
    return part(s, font=ARIAL)


def greek(s):
    return part(s, font=SYMBOL)


def bold(s):
    return part(s, font=ARIAL, weight="bold")


def sub(s):
    return part(s, font=ARIAL, size=SUB_SIZE, kind="sub")


def sup(s):
    return part(s, font=ARIAL, size=SUB_SIZE, kind="sup")


def save_svg(name, parts, width=None, height=22, x0=0.0):
    x = x0
    body = ""

    for k, p in enumerate(parts):
        if p["kind"] == "sub":
            y = BASE_Y + SUB_DY
        elif p["kind"] == "sup":
            y = BASE_Y + SUP_DY
        else:
            y = BASE_Y

        body += text_svg(
            x=x,
            y=y,
            content=p["content"],
            font=p["font"],
            size=p["size"],
            weight=p["weight"],
            label=f"{name}_part_{k}",
        )

        x += estimate_width(p["content"], p["size"]) + 0.4

    if width is None:
        width = x + 3

    svg = f'''<?xml version="1.0" encoding="UTF-8" standalone="no"?>
<svg xmlns="http://www.w3.org/2000/svg"
     xmlns:inkscape="http://www.inkscape.org/namespaces/inkscape"
     width="{width:.1f}pt"
     height="{height:.1f}pt"
     viewBox="0 0 {width:.1f} {height:.1f}">
{body}</svg>
'''
    (OUTDIR / name).write_text(svg, encoding="utf-8")


# η_ij = μ_i + γ_j − ||z_i − w_j||²
eq1 = [
    greek("η"), sub("ij"),
    latin(" = "),
    greek("μ"), sub("i"),
    latin(" + "),
    greek("γ"), sub("j"),
    latin(" − ||z"), sub("i"),
    latin(" − w"), sub("j"),
    latin("||"), sup("2"),
]
save_svg("scene_eta.svg", eq1)


# λ_ij = exp(η_ij), π_ij = σ(αη_ij)
eq2 = [
    greek("λ"), sub("ij"),
    latin(" = exp("),
    greek("η"), sub("ij"),
    latin("),   "),
    greek("π"), sub("ij"),
    latin(" = "),
    greek("σ"),
    latin("("),
    greek("α"),
    greek("η"), sub("ij"),
    latin(")"),
]
save_svg("scene_lambda_pi.svg", eq2)


# x_ij ~ Bernoulli(π_ij) · Poisson_{>0}(λ_ij)
eq3 = [
    latin("x"), sub("ij"),
    latin(" ~ Bernoulli("),
    greek("π"), sub("ij"),
    latin(") · Poisson"), sub(">0"),
    latin("("),
    greek("λ"), sub("ij"),
    latin(")"),
]
save_svg("scene_zip.svg", eq3)


# edge weight_ij = x_ij
eq_edge_weight = [
    latin("edge weight"), sub("ij"),
    latin(" = "),
    latin("x"), sub("ij"),
]
save_svg("scene_edge_weight.svg", eq_edge_weight)


# π_ij = σ(α(ψ_i + ω_j − ||z_i − w_j||_2))
eq_pi = [
    greek("π"), sub("ij"),
    latin(" = "),
    greek("σ"),
    latin("("),
    greek("α"),
    latin("("),
    greek("ψ"), sub("i"),
    latin(" + "),
    greek("ω"), sub("j"),
    latin(" − ||"),
    bold("z"), sub("i"),
    latin(" − "),
    bold("w"), sub("j"),
    latin("||"), sub("2"),
    latin("))"),
]
save_svg("scene_pi.svg", eq_pi)


# λ_ij = exp(ψ_i + ω_j − ||z_i − w_j||_2)
eq_lambda = [
    greek("λ"), sub("ij"),
    latin(" = exp("),
    greek("ψ"), sub("i"),
    latin(" + "),
    greek("ω"), sub("j"),
    latin(" − ||"),
    bold("z"), sub("i"),
    latin(" − "),
    bold("w"), sub("j"),
    latin("||"), sub("2"),
    latin(")"),
]
save_svg("scene_lambda.svg", eq_lambda)


print(f"Saved editable SVGs to: {OUTDIR.resolve()}")
```
